# Downloading Models

```python
from pathlib import Path
from huggingface_hub import login, snapshot_download

# Needed for the gated TookaBERT repo; paste your HF token when prompted.
login()

base_dir = Path("models/pretrained")
base_dir.mkdir(parents=True, exist_ok=True)

repos = [
    "PartAI/TookaBERT-Base",
    "HooshvareLab/roberta-fa-zwnj-base",
    "microsoft/mdeberta-v3-base",
]

for repo_id in repos:
    local_path = snapshot_download(
        repo_id=repo_id,
        local_dir=base_dir / repo_id.split("/")[-1],
        ignore_patterns=["*.msgpack", "*.h5", "*.ot", "*.onnx"],
    )
    print(f"Downloaded {repo_id} -> {local_path}")
```
